```python
import os
import numpy as np
import matplotlib.pyplot as plt
from Bio.PDB import PDBParser
import pandas as pd

def extract_LDDT(pdb_file):
    """ Extracts LDDT (or pLDDT) scores from a given PDB file. """
    parser = PDBParser(QUIET=True)
    structure = parser.get_structure('model', pdb_file)
    
    lddt_scores = []
    for model in structure:
        for chain in model:
            for residue in chain:
                for atom in residue:
                    lddt_scores.append(atom.get_bfactor())  # pLDDT values are stored in the B-factor column
                    break  # Only take the first atom's B-factor per residue
    
    return lddt_scores

def process_pdb_files(directory):
    """ Processes all PDB files in the given directory and extracts LDDT scores. """
    all_scores = {}

    for pdb_file in os.listdir(directory):
        if pdb_file.endswith('.pdb'):
            pdb_path = os.path.join(directory, pdb_file)
            lddt_scores = extract_LDDT(pdb_path)
            if lddt_scores:  # Ensure there's data
                all_scores[pdb_file] = np.array(lddt_scores)
    
    return all_scores

def plot_lddt_distribution(lddt_data):
    """ Plots the LDDT distribution with median markers and variability. """
    
    # Compute statistics
    file_names = list(lddt_data.keys())
    means = [np.mean(lddt_data[file]) for file in file_names]
    medians = [np.median(lddt_data[file]) for file in file_names]
    stds = [np.std(lddt_data[file]) for file in file_names]

    # Sort by mean LDDT
    sorted_indices = np.argsort(means)
    sorted_means = np.array(means)[sorted_indices]
    sorted_medians = np.array(medians)[sorted_indices]
    sorted_stds = np.array(stds)[sorted_indices]
    sorted_files = np.array(file_names)[sorted_indices]

    # Plot
    plt.figure(figsize=(12, 6))
    plt.errorbar(range(len(sorted_files)), sorted_means, yerr=sorted_stds, fmt='o', color='black', alpha=0.5, label="Mean ± Std Dev")
    plt.scatter(range(len(sorted_files)), sorted_medians, color='red', s=15, label="Median LDDT")
    
    plt.xlabel("Targets (Ordered by Mean LDDT)")
    plt.ylabel("LDDT Score")
    plt.title("Distribution of LDDT Scores Across Samples")
    plt.legend()
    plt.grid(True)
    plt.show()

# Specify your PDB directory
pdb_directory = "folds/mergedExotoxins_folds"

# Extract LDDT scores
lddt_scores = process_pdb_files(pdb_directory)

# Plot distribution
if lddt_scores:
    plot_lddt_distribution(lddt_scores)
else:
    print("No valid PDB files found in the directory.")
```

```python
# Required imports
import os
import glob
import numpy as np
import matplotlib.pyplot as plt
from Bio.PDB import PDBParser, Superimposer
from scipy.spatial import distance

# Ensure plots display inline
%matplotlib inline
```

```python
def compute_lddt(pred_pdb_path, ref_pdb_path, cutoff=15.0, bins=None):
    """
    Computes LDDT score between predicted and reference PDB files.
    
    Parameters:
        pred_pdb_path (str): Path to the predicted PDB file.
        ref_pdb_path (str): Path to the reference PDB file.
        cutoff (float): Distance cutoff in Angstroms.
        bins (list): List of distance thresholds for LDDT calculation.
        
    Returns:
        lddt_scores (list): LDDT scores per residue.
    """
    if bins is None:
        bins = [0.5, 1.0, 2.0, 4.0, 8.0]
    
    parser = PDBParser(QUIET=True)
    pred_structure = parser.get_structure('pred', pred_pdb_path)
    ref_structure = parser.get_structure('ref', ref_pdb_path)
    
    pred_model = next(pred_structure.get_models())
    ref_model = next(ref_structure.get_models())
    
    pred_chain = next(pred_model.get_chains())
    ref_chain = next(ref_model.get_chains())
    
    pred_ca = [res for res in pred_chain if 'CA' in res]
    ref_ca = [res for res in ref_chain if 'CA' in res]
    
    min_len = min(len(pred_ca), len(ref_ca))
    pred_ca = pred_ca[:min_len]
    ref_ca = ref_ca[:min_len]
    
    pred_coords = np.array([res['CA'].get_coord() for res in pred_ca])
    ref_coords = np.array([res['CA'].get_coord() for res in ref_ca])
    
    sup = Superimposer()
    sup.set_atoms([res['CA'] for res in ref_ca], [res['CA'] for res in pred_ca])
    sup.apply(pred_coords)
    
    lddt_scores = []
    for i in range(min_len):
        ref_res = ref_coords[i]
        pred_res = pred_coords[i]
        ref_dists = distance.cdist([ref_res], ref_coords, 'euclidean')[0]
        pred_dists = distance.cdist([pred_res], pred_coords, 'euclidean')[0]
        mask = (ref_dists <= cutoff) & (np.arange(min_len) != i)
        diff = np.abs(ref_dists[mask] - pred_dists[mask])
        
        score = 0.0
        for bin_threshold in bins:
            score += np.sum(diff <= bin_threshold)
        lddt = score / (len(bins) * np.sum(mask))
        lddt_scores.append(lddt * 100)
    return lddt_scores
```

```python
def plot_pretty_lddt_distribution(lddt_data):
    """
    Plots a visually appealing LDDT distribution.
    
    Parameters:
        lddt_data (dict): Dictionary where keys are file names and values are lists of LDDT scores.
    """
    file_names = list(lddt_data.keys())
    means = [np.mean(lddt_data[file]) for file in file_names]
    medians = [np.median(lddt_data[file]) for file in file_names]
    stds = [np.std(lddt_data[file]) for file in file_names]

    sorted_indices = np.argsort(means)
    sorted_means = np.array(means)[sorted_indices]
    sorted_medians = np.array(medians)[sorted_indices]
    sorted_stds = np.array(stds)[sorted_indices]
    sorted_files = np.array(file_names)[sorted_indices]

    plt.figure(figsize=(12, 6))
    plt.plot(range(len(sorted_files)), sorted_means, color='black', linewidth=2, label="Mean LDDT")
    plt.errorbar(range(len(sorted_files)), sorted_means, yerr=sorted_stds, fmt='o', 
                 color='gray', alpha=0.3, capsize=2, label="Mean ± Std Dev")
    plt.scatter(range(len(sorted_files)), sorted_medians, color='red', s=8, label="Median LDDT")
    plt.xlabel("Targets (Ordered by Mean LDDT)", fontsize=12)
    plt.ylabel("LDDT Score", fontsize=12)
    plt.title("Distribution of LDDT Scores Across Samples", fontsize=14, fontweight='bold')
    plt.legend()
    plt.grid(True, linestyle="--", alpha=0.5)
    plt.tight_layout()
    plt.show()
```

```python
# Define directories
pred_dir = "folds/mergedExotoxins_folds"  # Replace with the path to your predicted PDB files
ref_dir = "folds/controlProteins_folds"  # Replace with the path to your reference PDB files

# Get list of PDB files
pred_pdb_files = glob.glob(os.path.join(pred_dir, "*.pdb"))
ref_pdb_files = glob.glob(os.path.join(ref_dir, "*.pdb"))

# Create a mapping from file name (without extension) to path
pred_dict = {os.path.splitext(os.path.basename(f))[0]: f for f in pred_pdb_files}
ref_dict = {os.path.splitext(os.path.basename(f))[0]: f for f in ref_pdb_files}

# Find common files
common_files = set(pred_dict.keys()).intersection(set(ref_dict.keys()))

# Compute LDDT scores
lddt_data = {}
for file_id in common_files:
    pred_path = pred_dict[file_id]
    ref_path = ref_dict[file_id]
    print(f"Computing LDDT for {file_id}...")
    try:
        lddt_scores = compute_lddt(pred_path, ref_path)
        lddt_data[file_id] = lddt_scores
    except Exception as e:
        print(f"Error computing LDDT for {file_id}: {e}")

# Check if any LDDT data was computed
if lddt_data:
    plot_pretty_lddt_distribution(lddt_data)
else:
    print("No LDDT data computed.")
```
